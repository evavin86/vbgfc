```python
A = float(input())
B = float(input())
C = float(input())
count_A = 0
while A >= C:
    A -= C
    count_A += 1
count_B = 0
while B >= C:
    B -= C
    count_B += 1
total_squares = 0
while count_A > 0:
    total_squares += count_B
    count_A -= 1
print(total_squares)
```
